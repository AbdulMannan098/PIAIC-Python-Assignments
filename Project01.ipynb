# LangChain and Google Gemini Flash 2.0 Integration
This notebook demonstrates the integration of LangChain with Google Gemini Flash 2.0 to answer user-defined questions. We experiment with various prompt templates and model parameters to optimize responses.

## Key Features
- Dynamic prompt templates
- Adjustable parameters: temperature, max tokens
- User-interactive widgets

## 2. Setting Up the Environment

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
```

```python
from google.colab import userdata
api_key=userdata.get('GOOGLE_API_KEY_1')
```

## 3. Creating an LLM chain using LangChain and Defining a prompt template.

```python
llm = ChatGoogleGenerativeAI(
    api_key=api_key,
    model="gemini-2.0-flash-exp",
    temperature=0.7
)
result = llm.invoke("Write a Note about LangChain")
print(result.content)
```

```python
prompt_template = PromptTemplate(
    input_variables=["question"],
    template="You are an intelligent assistant. Answer the following question concisely and accurately: {question}"
)

chain = LLMChain(llm=llm, prompt=prompt_template)
```

## 4. Running The Chain With User-defined Questions.

```python
# List of user-defined questions
questions = [
    "What is the capital of France?",
    "Explain the significance of LangChain in AI development.",
    "How does the Gemini Flash 2.0 model differ from traditional LLMs?"
]

# Run the chain and display responses
for question in questions:
    response = chain.run(question)
    print(f"Question: {question}\nResponse: {response}\n")
```

## 5. Experimenting With Different Prompt Templates

```python
# Experiment with different parameters
llm_high_temperature = ChatGoogleGenerativeAI(api_key=api_key,model="gemini-1.5-pro", temperature=1.0, max_tokens=200)
chain_high_temp = LLMChain(llm=llm_high_temperature, prompt=prompt_template)

# Test a new prompt template
creative_prompt_template = PromptTemplate(
    input_variables=["question"],
    template="Be creative and provide an engaging answer to this question: {question}"
)
chain_creative = LLMChain(llm=llm, prompt=creative_prompt_template)

# Display comparisons
print("Standard Configuration Responses:")
for question in questions:
    response = chain.run(question)
    print(f"Q: {question}\nA: {response}\n")

print("High Temperature Responses:")
for question in questions:
    response = chain_high_temp.run(question)
    print(f"Q: {question}\nA: {response}\n")

print("Creative Prompt Responses:")
for question in questions:
    response = chain_creative.run(question)
    print(f"Q: {question}\nA: {response}\n")
```

## 7. User-interactive widgets (Additional)

```python
from ipywidgets import interact

# Interactive widget to adjust parameters
def interactive_chain(question, temperature=0.7, max_tokens=150):
    llm_dynamic = ChatGoogleGenerativeAI(
        api_key=api_key,
        model="gemini-2.0-flash-exp", temperature=temperature, max_tokens=max_tokens
    )
    chain_dynamic = LLMChain(llm=llm_dynamic, prompt=prompt_template)
    response = chain_dynamic.run(question)
    return response

interact(interactive_chain, question="Type your question here...", temperature=(0, 1, 0.1), max_tokens=(50, 500, 50))
```
